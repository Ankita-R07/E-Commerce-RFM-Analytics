```python
# 1. Load Data

import pandas as pd

print("\n1. LOAD DATA")

file_path = "C:\\Users\\Admin\\Downloads\\Portfolio Projects\\E-Commerce-RFM-Analytics\\data\\online_retail_II.xlsx"

excel_file = pd.ExcelFile(file_path)

print("\nAvailable Worksheets:")
print(excel_file.sheet_names)

sheet1 = excel_file.parse("Year 2009-2010")
sheet2 = excel_file.parse("Year 2010-2011")

sheets = {
    "2009-2010": sheet1,
    "2010-2011": sheet2
}
```

```python
# 2. Dataset Structure

print("\n2. DATASET STRUCTURE")

for year, df in sheets.items():

    print(f"\n--- {year} ---")

    print("Shape : ", df.shape)
    print("Columns : ", df.columns.tolist())
```

```python
# 3. Data Completeness

print("\n3. DATA COMPLETENESS")

for year, df in sheets.items():

    print(f"\n--- {year} ---")

    print("\nMissing Values : ")
    print(df.isnull().sum())

    missing_customer_pct = (df["Customer ID"].isna().mean() * 100)

    print(
        "\nMissing Customer ID : ",
        round(missing_customer_pct, 2),
        "%"
    )
```

```python
# 4. Data Types

print("\n4. DATA TYPES")

for year, df in sheets.items():

    print(f"\n--- {year} ---")
    print(df.dtypes)
```

```python
# 5. Transaction Validity

print("\n5. TRANSACTION VALIDITY")

for year, df in sheets.items():

    zero_quantity = (df["Quantity"] == 0)
    negative_quantity = (df["Quantity"] < 0)

    zero_price = (df["Price"] == 0)
    negative_price = (df["Price"] < 0)

    print(f"\n--- {year} ---")

    print("Zero Quantity Rows : ", zero_quantity.sum())

    print("Negative Quantity : ", negative_quantity.sum())

    print("Zero Price Rows : ", zero_price.sum())

    print("Negative Price Rows : ", negative_price.sum())

    # Inspect Negative Price Rows
    if negative_price.sum() > 0:

        print("\nNegative Price Records : ")
        print(df.loc[negative_price])
```

```python
# 6. Cancellation / Return Analysis

print("\n6. CANCELLATION / RETURN ANALYSIS")

for year, df in sheets.items():

    cancellation = df["Invoice"].str.startswith("C", na = False)

    negative_without_cancellation = ((df["Quantity"] < 0) & (~cancellation))

    print(f"\n--- {year} ---")

    print("Cancellation Transaction Rows : ", cancellation.sum())

    print(
        "Unique Cancellation Invoices  : ",
        df.loc[cancellation, "Invoice"].nunique()
    )

    print(
        "Negative Qty without Cancellation :",
        negative_without_cancellation.sum()
    )
```

```python
# 7. Non-Standard Records

print("\n7. NON-STANDARD RECORDS")

for year, df in sheets.items():

    manual = df["StockCode"] == "M"
    test = df["StockCode"] == "TEST001"

    print(f"\n--- {year} ---")

    print("Manual Records : ", manual.sum())

    print("TEST001 Records : ", test.sum())
```

```python
# 8. Duplicate & Uniqueness Check

print("\n8. DUPLICATE & UNIQUENESS CHECK")

for year, df in sheets.items():

    duplicates = df.duplicated()

    print(f"\n--- {year} ---")

    print("Exact Duplicate Rows : ", duplicates.sum())

    print("Unique Customers : ", df["Customer ID"].nunique())

    print("Unique Invoices : ", df["Invoice"].nunique())

    print("\nDuplicate Sample : ")
    print(df.loc[duplicates].head(5))
```

```python
# 9. Date Coverage

print("\n9. DATE COVERAGE")

for year, df in sheets.items():

    print(f"\n--- {year} ---")

    print("Start : ", df["InvoiceDate"].min())

    print("End : ", df["InvoiceDate"].max())
```

```python
# 10. Cross-Sheet Consistency

print("\n10. CROSS-SHEET CONSISTENCY")

common_invoices = set(sheet1["Invoice"]).intersection(
    set(sheet2["Invoice"]))

print(
    "\nInvoices appearing in both sheets : ",
    len(common_invoices)
)

# Date overlap

start_1 = sheet1["InvoiceDate"].min()
end_1 = sheet1["InvoiceDate"].max()

start_2 = sheet2["InvoiceDate"].min()
end_2 = sheet2["InvoiceDate"].max()

overlap_start = max(start_1, start_2)
overlap_end = min(end_1, end_2)

print("\nDate overlap : ")

if overlap_start <= overlap_end:

    print("Date overlap exists : YES")
    print("Overlap Start : ", overlap_start)
    print("Overlap End : ", overlap_end)

else:

    print("Date overlap exists : NO")
```

```python
# 11. Final Audit Summary

print("\n" + "_" * 60)
print("\n11. FINAL AUDIT SUMMARY")
print("_" * 60)

print("\n")
audit_results = []

for year, df in sheets.items():

    cancellation = df["Invoice"].str.startswith("C", na = False)

    negative_without_cancellation = (
        (df["Quantity"] < 0)
        & (~cancellation)
    )

    audit_results.append({

        "Year": year,

        "Rows": len(df),

        "Missing Customer ID": (
            df["Customer ID"].isna().sum()
        ),

        "Missing Customer ID %": round(
            df["Customer ID"].isna().mean() * 100,
            2
        ),

        "Missing Description": (
            df["Description"].isna().sum()
        ),

        "Zero Quantity": (
            (df["Quantity"] == 0).sum()
        ),

        "Negative Quantity": (
            (df["Quantity"] < 0).sum()
        ),

        "Zero Price": (
            (df["Price"] == 0).sum()
        ),

        "Negative Price": (
            (df["Price"] < 0).sum()
        ),

        "Cancellation Rows": (
            cancellation.sum()
        ),

        "Negative Qty without Cancellation": (
            negative_without_cancellation.sum()
        ),

        "Manual Records": (
            (df["StockCode"] == "M").sum()
        ),

        "TEST001 Records": (
            (df["StockCode"] == "TEST001").sum()
        ),

        "Exact Duplicates": (
            df.duplicated().sum()
        ),

        "Unique Customers": (
            df["Customer ID"].nunique()
        ),

        "Unique Invoices": (
            df["Invoice"].nunique()
        )
    })


audit_summary = pd.DataFrame(audit_results)

print(audit_summary)
```
